```python
import string

def custom_encrypt_message(message):
    encrypted_chars = []
    message = message.lower() # Convert input to lowercase

    char_map = {}
    alphabet = string.ascii_lowercase
    for i in range(len(alphabet)):
        char_map[alphabet[i]] = alphabet[len(alphabet) - 1 - i]

    for char in message:
        if char in char_map:
            encrypted_chars.append(char_map[char])
        elif char.isspace(): # If it's a whitespace character
            encrypted_chars.append('$')
        else: # Keep punctuation and other characters unchanged
            encrypted_chars.append(char)

    return ''.join(encrypted_chars)


# Input Sentence provided by the user
input_sentence_to_encrypt = "I want to become a Data Scientist."

# Encrypt the sentence using the new function
encrypted_output = custom_encrypt_message(input_sentence_to_encrypt)

# Print the original (lowercase) and encrypted sentence
print(f"Original (lowercase): {input_sentence_to_encrypt.lower()}")
print(f"Encrypted: {encrypted_output}")

```
